```python
import requests
from bs4 import BeautifulSoup
import numpy as np
import time
product_name=[]
rating=[]
features= []
reviews= []
price=[]
discount=[]
pagenum=[]
for i in range(1,25):
    start =time.time()
    URL = 'https://www.flipkart.com/search?q=tv+smart+tv{}'.format(i)
    headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
}
    page = requests.get(url, headers=headers)
    pagecontent=page.text
    soup=BeautifulSoup(pagecontent)
    
    for x in soup.find_all('div',attrs={'class':'ZFwe0M row'}):
        
        product_name1= x.find('div',attrs={'class':'RG5Slk'})
        rating1= x.find('div',attrs={'class':'MKiFS6'})
        features1= x.find('li',attrs={'class':'DTBslk'})
        reviews1= x.find('span',attrs={'class':'PvbNMB'})
        price1=x.find('div',attrs={'class':'hZ3P6w DeU9vF'})
        discount1=x.find('div',attrs={'class':'kRYCnD gxR4EY'})
    

        if product_name1 is None:
            product_name.append(np.nan)
        else:
            product_name.append(product_name1.text)

        if rating1 is None:
            rating.append(np.nan)
        else:
            rating.append(rating1.text)

        if features1 is None:
            features.append(np.nan)
        else:
            features.append(features1.text)

        if reviews1 is None:
            reviews.append(np.nan)
        else:
            reviews.append(reviews1.text)

        if  price1 is None:
             price.append(np.nan)
        else:
             price.append(price1.text)

        if discount1 is None:
            discount.append(np.nan)
        else:
            discount.append(discount1.text)

        pagenum.append(i)
    end=time.time()
    
    print('page {} is completed in {} seconds'.format(i,end-start))
```

```python
# for price column to remove (₹ and ,)
# convert to integer
df["Price"] = df["Price"].str.replace("₹", "")
df["Price"] = df["Price"].str.replace(",", "")
df["Price"] = df["Price"].astype(int)
df
```

```python
# converting rating to float
df["Rating"] = pd.to_numeric(df["Rating"], errors="coerce")
df
```

```python
# renaming of Rating column to rating point
df.rename(columns={'Rating': 'Rating_point'}, inplace=True)
df.head()
```

```python
# data manipulation
# creating new columns as separate ratings and reviews
# converting to int
df["Ratings"] = df["Reviews"].str.extract(r'([\d,]+)\sRatings')
df["Ratings"] = df["Ratings"].str.replace(",", "")
df["Ratings"] = df["Ratings"].astype(int)
df.head()
```

```python
# creating of new col reviews
df["reviews"] = df["Reviews"].str.extract(r'([\d,]+)\sReviews')
df["reviews"] = df["reviews"].str.replace(",", "")
df["reviews"] = df["reviews"].astype(int)
df.head()
```

```python
# for converting Reviews to ratings and reviews
df.rename(columns={'Reviews': 'Ratings and Reviews'}, inplace=True)
df.head()
```

```python
# to remove & symbol in df
df['Ratings and Reviews']=df['Ratings and Reviews'].str.replace('&','')
df.head()
```

# bivariate analysis 

```python
# #Top 5 highest-rated TV's
# horizontal bar chart
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv('TV_data.csv')
top5=df.groupby('BrandName')['Ratings'].mean().sort_values(ascending=False).head(5)
top5
```

```python
sns.barplot(x=top5.index, y=top5.values)
plt.title('Top 5 highest-rated TVs brand')
plt.xlabel('BrandName')
plt.ylabel('Ratings')
plt.show()

#description :
#1. The bar chart represents the Top 5 highest-rated TV brands based on customer ratings data. Among all brands, (Thomson)  shows the highest number of ratings, indicating strong customer engagement and popularity in the dataset

#2. XIAOMI and Foxsky follow as the second and third highest-rated brands, demonstrating competitive market performance and significant customer feedback compared to other brands

#3.This visualization helps me to  identify brand performance trends and customer preference patterns, making it easier to compare rating distribution across major TV brands
```

```python
# avg price for brand
avg_price=df.groupby('BrandName')['Price'].mean().sort_values(ascending=False)
sns.barplot(x=avg_price.index, y=avg_price.values)
plt.title('avg price for brand')
plt.xticks(rotation=90)
plt.xlabel('BrandName')
plt.ylabel('Price')
plt.show()


#description :
#1. The bar chart illustrates the average price distribution across different TV brands. Among all brands, (Philips) has the highest average price, indicating its positioning in the premium market segment.
#2. XIAOMI and Lumino follow as the second and third highest-priced brands, suggesting a mid-to-premium pricing strategy
#3. this helps me to understand brand-wise pricing strategies and market segmentation
#4. to know the present market pricing from this 
```

```python
# which brand has more products in market
brand_count = df['BrandName'].value_counts()
brand_count
```

```python
colors=sns.color_palette('bright')[0:8]
brand_count.plot(kind='pie', autopct='%1.1f%%', startangle=90)
plt.title('the brands which are having more products in market')
plt.show()

#description:
#1. The pie chart illustrates the distribution of TV brands based on the number of products available in the market. It highlights each brand’s percentage contribution to the total product count.
#2. Reliance holds the largest market share with approximately 25%, indicating the highest number of available products among all brands.
#3. Uniboom (14.1%), VW (10.9%), and Lumio (10.1%) also show significant product presence, suggesting strong market participation
#4. This visualization helps me to  analyze brand-wise product availability and market penetration
```

```python
# Price vs Rating Comparison
plt.scatter(df['Price'], df['Ratings'])
plt.title('Price vs Rating Comparison')
plt.xlabel("Price")
plt.ylabel("Ratings")
plt.show()

#description:
#1.The scatter plot represents the relationship between TV price and the number of customer ratings. Each data point corresponds to a product, showing how ratings vary across different price levels.
#2. The majority of products are clustered in the lower price range (below ₹10,000), indicating that budget-friendly customer ratings.
#3. A few high-priced TVs (₹30,000–₹60,000 range) are visible; however, they generally show fewer ratings
#4. he visualization indicates no strong positive correlation between price and number of ratings, meaning higher-priced TVs do not necessarily receive more customer reviews
```

# univariate analysis

```python
# price distribution
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.histplot(data=df, x='Price', bins=20)
plt.title('Price Analysis')
plt.show()

#description:
#1. The histogram represents the distribution of TV prices in the dataset. It shows how frequently products fall within different price ranges.
#2. The majority of TVs are concentrated in the lower price range (approximately ₹5,000 – ₹10,000), indicating that the market is largely dominated by budget-friendly models.
#3.  A smaller number of products fall within the mid-range segment (₹25,000 – ₹45,000), suggesting moderate availability of premium or advanced feature TVs.
```

```python
# rating point analysis 
plt.hist(df['Rating_point'], bins=10)
plt.title('rating analysis')
plt.show()

#description:
#1. The ratings predominantly cluster around 4.0, indicating that most observations fall within a high-rating range, suggesting generally positive evaluations
#2. The distribution appears slightly concentrated between 3.7 and 4.2, with fewer instances at the lower (around 3.5) and higher (above 4.3) extremes
#3.  lower rating have less count it means , customers are always look into rating for better product
```

# multivariate analysis

```python
# correlation heatmap b/w price , ratings , reviews
sns.heatmap(df[['Price','Ratings','reviews']].corr(),annot=True)
plt.title('correlation heatmap analysis b/w price , ratings , reviews')
plt.show()

#description:
#1. a correlation heatmap tells the relationship between Price, Ratings, and Reviews
#2. A strong positive correlation (0.97) is observed between Ratings and Reviews, indicating that items with higher ratings tend to receive a greater number of reviews.
#3. Price shows a weak negative correlation with both Ratings (-0.12) and Reviews (-0.18), suggesting that price has minimal influence on customer ratings and review count
```
